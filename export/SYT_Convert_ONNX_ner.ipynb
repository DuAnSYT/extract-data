```python
from optimum.onnxruntime import ORTQuantizer
from optimum.onnxruntime.configuration import AutoQuantizationConfig

# 1. Định nghĩa đường dẫn
onnx_model_dir = "onnx_ner_output" # Folder chứa file model.onnx gốc (FP32)
output_dir = "visobert_onnx_quantized" # Folder sẽ chứa model INT8

# 2. Load Quantizer
quantizer = ORTQuantizer.from_pretrained(onnx_model_dir, file_name="model.onnx")

# 3. Tạo Config Dynamic
# is_static=False chính là kích hoạt chế độ Dynamic
# per_channel=True: Giúp giữ độ chính xác cao hơn (tuy hơi chậm hơn xíu xiu so với per_tensor)
# Chọn avx512_vnni nếu server Intel đời mới, hoặc avx2 (an toàn cho hầu hết server Intel/AMD)
qconfig = AutoQuantizationConfig.avx2(is_static=False, per_channel=True)

# 4. Thực hiện Quantize
quantizer.quantize(save_dir=output_dir, quantization_config=qconfig)
```
